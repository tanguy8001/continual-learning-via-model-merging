```python
import numpy as np

a = np.array([[1, 2],
              [3, 4]])
b = np.array([[5, 6],
              [7, 8]])

print(a * b)
print(a @ b)

a = np.matrix([[1, 2],
              [3, 4]])
b = np.matrix([[5, 6],
              [7, 8]])

print(a * b)
print(a @ b)
```
